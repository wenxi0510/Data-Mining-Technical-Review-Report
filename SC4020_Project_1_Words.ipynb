```python
import gensim.downloader as api
from sklearn.neighbors import KDTree, BallTree
from annoy import AnnoyIndex
import hnswlib
import numpy as np
import matplotlib.pyplot as plt
from tqdm import tqdm
import time
import pandas as pd
from collections import defaultdict
from sklearn.decomposition import PCA
import warnings
from itertools import combinations

warnings.filterwarnings('ignore')

MODEL_NAME = "glove-wiki-gigaword-50"
DATASET_SIZE = 1000
k_values = [2, 5, 10, 20, 30, 40, 50,100,500,1000]

num_query_samples = 50
annoy_trees = 10
hnsw_ef = 50
dense_lsh_planes = 10
sizes_to_test = [200, 400, 600, 800, 1000]
random_seed = 42
np.random.seed(random_seed)

algorithms = ['Brute-Force', 'KD-Tree', 'Ball-Tree', 'Annoy', 'HNSW', 'LSH']

try:
    with open("20k.txt", "r", encoding="utf-8") as f:
        words = [line.strip() for line in f if line.strip()]
except FileNotFoundError:
    words = ['computer', 'software', 'technology', 'science', 'math', 'keyboard', 'mouse', 'program', 'data', 'algorithm']
    print("WARNING: '20k.txt' not found. Using fallback list.")

print("Loading GloVe embeddings...")
model = api.load(MODEL_NAME)
words = words[:DATASET_SIZE]
words_in_model = [w for w in words if w in model.key_to_index]
embeddings = np.array([model[w] for w in words_in_model])

DIMENSION = embeddings.shape[1]
NUM_ELEMENTS = len(words_in_model)
query_word = 'computer' if 'computer' in words_in_model else words_in_model[0]
query_vec = model[query_word].reshape(1, -1)
query_idx = words_in_model.index(query_word)

print(f"Using {NUM_ELEMENTS} words, embedding dim = {DIMENSION}. Query = '{query_word}'")

pca = PCA(n_components=2)
emb_2d = pca.fit_transform(embeddings)

class DenseVectorLSH:
    def __init__(self, dim, num_planes=24):
        self.dim = dim
        self.num_planes = num_planes
        self.planes = np.random.randn(num_planes, dim)
        self.hash_table = defaultdict(list)

    def _hash(self, vector):
        v = vector.flatten()
        bits = (self.planes.dot(v) > 0).astype(int)
        return tuple(bits.tolist())

    def insert(self, idx, vector):
        h = self._hash(vector)
        self.hash_table[h].append(idx)

    def _hamming_neighbors(self, h, radius=2):
        bits = list(h)
        n = len(bits)
        for r in range(1, radius + 1):
            for comb in combinations(range(n), r):
                nb = bits.copy()
                for i in comb:
                    nb[i] = 1 - nb[i]
                yield tuple(nb)

    def query(self, vector, top_k=10):
        h = self._hash(vector)
        candidates = list(self.hash_table.get(h, []))

        if len(candidates) < 1:
            for h2 in self._hamming_neighbors(h, radius=2):
                if h2 in self.hash_table:
                    candidates.extend(self.hash_table[h2])
                if len(candidates) >= max(5, top_k):
                    break

        if len(candidates) == 0:
            sample_size = min(200, len(embeddings))
            if sample_size == 0:
                return []
            sampled = np.random.choice(len(embeddings), size=sample_size, replace=False)
            candidates = list(sampled)

        dists = np.array([np.linalg.norm(vector - embeddings[i]) for i in candidates])
        order = np.argsort(dists)
        result = [int(candidates[i]) for i in order[:top_k]]
        return result

def get_neighbors_indices(k, index_obj, alg, qv, qidx, subset):
    k_eff = min(k, max(1, len(subset)))
    if alg == 'Brute-Force' or index_obj is None:
        dists = np.linalg.norm(subset - qv, axis=1)
        ordered = [i for i in np.argsort(dists) if i != qidx]
        return ordered[:k_eff]
    elif alg in ['KD-Tree', 'Ball-Tree']:
        k_tree = min(k_eff, len(subset))
        _, ind = index_obj.query(qv, k=k_tree)
        ordered = [i for i in ind[0] if i != qidx]
        return ordered[:k_eff]
    elif alg == 'Annoy':
        got = index_obj.get_nns_by_vector(qv.flatten(), k_eff)
        ordered = [i for i in got if i != qidx]
        return ordered[:k_eff]
    elif alg == 'HNSW':
        labels, _ = index_obj.knn_query(qv, k=k_eff)
        ordered = [i for i in labels[0] if i != qidx]
        return ordered[:k_eff]
    elif alg == 'LSH':
        got = index_obj.query(qv.flatten(), top_k=k_eff)
        ordered = [i for i in got if i != qidx]
        return ordered[:k_eff]
    return []

def precision_at_k(retrieved, relevant):
    if not retrieved: return 0.0
    return len(set(retrieved) & set(relevant)) / len(retrieved)
def recall_at_k(retrieved, relevant):
    if not relevant: return 0.0
    return len(set(retrieved) & set(relevant)) / len(relevant)
def mrr_at_k(retrieved, relevant):
    for rank, idx in enumerate(retrieved, start=1):
        if idx in relevant:
            return 1.0 / rank
    return 0.0
def map_at_k(retrieved, relevant):
    if not relevant: return 0.0
    hits = 0; sum_prec = 0.0
    for i, idx in enumerate(retrieved, 1):
        if idx in relevant:
            hits += 1
            sum_prec += hits / i
    return sum_prec / len(relevant)
def ndcg_at_k(retrieved, relevant):
    if not relevant: return 0.0
    dcg = sum((1.0 / np.log2(i + 2)) for i, idx in enumerate(retrieved) if idx in relevant)
    idcg = sum((1.0 / np.log2(i + 2)) for i in range(len(relevant)))
    return (dcg / idcg) if idcg > 0 else 0.0

alg_metrics = {a:{'precisions':[],'recalls':[],'mrrs':[],'maps':[],'ndcgs':[],
                   'query_times':[],'build_times':[],'index':None} for a in algorithms}

# KDTree
t = time.time(); kd = KDTree(embeddings); alg_metrics['KD-Tree']['index'] = kd; alg_metrics['KD-Tree']['build_times'].append(time.time()-t)

# BallTree
t = time.time(); bt = BallTree(embeddings); alg_metrics['Ball-Tree']['index'] = bt; alg_metrics['Ball-Tree']['build_times'].append(time.time()-t)

# Annoy
t = time.time(); an = AnnoyIndex(DIMENSION, 'angular')
for i, e in enumerate(embeddings): an.add_item(i, e)
an.build(annoy_trees)
alg_metrics['Annoy']['index'] = an; alg_metrics['Annoy']['build_times'].append(time.time()-t)

# HNSW
t = time.time(); h = hnswlib.Index(space='l2', dim=DIMENSION)
h.init_index(max_elements=NUM_ELEMENTS, ef_construction=100, M=16)
h.add_items(embeddings, np.arange(NUM_ELEMENTS)); h.set_ef(hnsw_ef)
alg_metrics['HNSW']['index'] = h; alg_metrics['HNSW']['build_times'].append(time.time()-t)

# LSH
t = time.time(); lsh = DenseVectorLSH(DIMENSION, num_planes=dense_lsh_planes)
for i, v in enumerate(embeddings): lsh.insert(i, v)
alg_metrics['LSH']['index'] = lsh; alg_metrics['LSH']['build_times'].append(time.time()-t)

# Brute Force
alg_metrics['Brute-Force']['index'] = None; alg_metrics['Brute-Force']['build_times'].append(0.0)

query_indices = np.random.choice(NUM_ELEMENTS, size=min(num_query_samples, NUM_ELEMENTS), replace=False)

for alg in tqdm(algorithms, desc="Evaluating algorithms"):
    index_obj = alg_metrics[alg]['index']
    for k in k_values:
        precisions, recalls, mrrs, maps, ndcgs, q_times = [], [], [], [], [], []
        for q in query_indices:
            qv = embeddings[q].reshape(1, -1)
            true_neighbors = get_neighbors_indices(k, None, 'Brute-Force', qv, q, embeddings)
            start = time.time()
            retrieved = get_neighbors_indices(k, index_obj, alg, qv, q, embeddings)
            elapsed = (time.time() - start) * 1000.0
            precisions.append(precision_at_k(retrieved, true_neighbors))
            recalls.append(recall_at_k(retrieved, true_neighbors))
            mrrs.append(mrr_at_k(retrieved, true_neighbors))
            maps.append(map_at_k(retrieved, true_neighbors))
            ndcgs.append(ndcg_at_k(retrieved, true_neighbors))
            q_times.append(elapsed)
        alg_metrics[alg]['precisions'].append(np.mean(precisions))
        alg_metrics[alg]['recalls'].append(np.mean(recalls))
        alg_metrics[alg]['mrrs'].append(np.mean(mrrs))
        alg_metrics[alg]['maps'].append(np.mean(maps))
        alg_metrics[alg]['ndcgs'].append(np.mean(ndcgs))
        alg_metrics[alg]['query_times'].append(np.mean(q_times))

def plot_pca_topk(alg, top_k=5):
    # safe top_k
    top_k = min(top_k, NUM_ELEMENTS - 1)
    idx_obj = alg_metrics[alg]['index']
    neighbors_idx = get_neighbors_indices(top_k+1, idx_obj, alg, query_vec, query_idx, embeddings)
    neighbors_idx = [i for i in neighbors_idx if i != query_idx][:top_k]
    plt.figure(figsize=(10,8))
    plt.scatter(emb_2d[:,0], emb_2d[:,1], alpha=0.25, s=40, color='lightgray')
    qx, qy = emb_2d[query_idx]
    plt.scatter(qx, qy, color='red', s=150, label=f'Query: {query_word}')
    if len(neighbors_idx) > 0:
        nx, ny = emb_2d[neighbors_idx,0], emb_2d[neighbors_idx,1]
        plt.scatter(nx, ny, color='blue', s=80, label=f'Top {len(neighbors_idx)} neighbors')
        for i in neighbors_idx:
            plt.plot([qx, emb_2d[i,0]], [qy, emb_2d[i,1]], color='green', linewidth=1.0, alpha=0.7)
            plt.text(emb_2d[i,0]+0.01, emb_2d[i,1]+0.01, words_in_model[i], fontsize=8)
    else:
        plt.text(qx, qy, " (no neighbors found)", fontsize=10, color='black', va='bottom')
    plt.title(f"2D PCA Projection - {alg} (Top {top_k})")
    plt.xlabel("PC1"); plt.ylabel("PC2"); plt.legend(); plt.grid(alpha=0.3); plt.show()

pca_top_k = 5
for alg in algorithms:
    plot_pca_topk(alg, top_k=pca_top_k)

def plot_metric(metric_name, ylabel, fmt="%.4f"):
    df = pd.DataFrame({alg: alg_metrics[alg][metric_name] for alg in algorithms}, index=k_values)
    print(f"\n=== {ylabel} vs K ===")
    display(df.round(6))
    plt.figure(figsize=(10,5))
    for alg in algorithms:
        plt.plot(k_values, df[alg], marker='o', label=alg)
    plt.xlabel("K"); plt.ylabel(ylabel); plt.title(f"{ylabel} vs K"); plt.legend(); plt.grid(True); plt.show()

plot_metric('precisions','Precision@K')
plot_metric('recalls','Recall@K')
plot_metric('mrrs','MRR@K')
plot_metric('maps','MAP@K')
plot_metric('ndcgs','nDCG@K')
plot_metric('query_times','Avg Query Time (ms)')

scaling_results = {alg:{'size':[],'precision':[],'recall':[],'time':[]} for alg in algorithms}

scale_k = 1000
for size in tqdm(sizes_to_test, desc="Scaling dataset size"):
    subset_emb = embeddings[:size]
    subset_indices = np.random.choice(size, size=min(20, size), replace=False)
    for alg in algorithms:
        if alg == 'Brute-Force':
            idx_obj = None
        elif alg == 'KD-Tree':
            idx_obj = KDTree(subset_emb)
        elif alg == 'Ball-Tree':
            idx_obj = BallTree(subset_emb)
        elif alg == 'Annoy':
            idx_obj = AnnoyIndex(DIMENSION, 'angular')
            for i,e in enumerate(subset_emb): idx_obj.add_item(i,e)
            idx_obj.build(annoy_trees)
        elif alg == 'HNSW':
            idx_obj = hnswlib.Index(space='l2',dim=DIMENSION)
            idx_obj.init_index(max_elements=size, ef_construction=100, M=16)
            idx_obj.add_items(subset_emb, np.arange(size)); idx_obj.set_ef(hnsw_ef)
        elif alg == 'LSH':
            idx_obj = DenseVectorLSH(DIMENSION, num_planes=dense_lsh_planes)
            for i,v in enumerate(subset_emb): idx_obj.insert(i,v)

        precs, recs, times = [], [], []
        for q in subset_indices:
            qv = subset_emb[q].reshape(1,-1)
            true_neighbors = get_neighbors_indices(scale_k, None, 'Brute-Force', qv, q, subset_emb)
            start = time.time()
            retrieved = get_neighbors_indices(scale_k, idx_obj, alg, qv, q, subset_emb)
            elapsed = (time.time()-start)*1000.0
            precs.append(precision_at_k(retrieved, true_neighbors))
            recs.append(recall_at_k(retrieved, true_neighbors))
            times.append(elapsed)
        scaling_results[alg]['size'].append(size)
        scaling_results[alg]['precision'].append(np.mean(precs))
        scaling_results[alg]['recall'].append(np.mean(recs))
        scaling_results[alg]['time'].append(np.mean(times))

def plot_scaling(metric, ylabel):
    df = pd.DataFrame({alg: scaling_results[alg][metric] for alg in algorithms}, index=sizes_to_test)
    print(f"\n=== {ylabel} vs Dataset Size ===")
    display(df.round(6))
    plt.figure(figsize=(10,5))
    for alg in algorithms:
        plt.plot(sizes_to_test, df[alg], marker='o', label=alg)
    plt.xlabel("Dataset Size"); plt.ylabel(ylabel); plt.title(f"{ylabel} vs Dataset Size"); plt.legend(); plt.grid(True); plt.show()

plot_scaling('precision','Precision@50')
plot_scaling('recall','Recall@50')
plot_scaling('time','Avg Query Time (ms)')
```
